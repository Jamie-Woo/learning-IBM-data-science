### `Practical Example`
Consider a study investigating the relationship between smoking status (smoker, non-smoker) and the incidence of lung disease (disease, no disease). The researcher collects data from 200 individuals and records the following information:

```python
import pandas as pd

data = {
    'Category': ['Smoker', 'Non-Smoker', 'Total'],
    'Disease': [50, 20, 70],
    'No Disease': [30, 100, 130],
    'Total': [80, 120, 200]
}

df = pd.DataFrame(data)
print(df)
```

#### step1 : Calculate Expected Frequencies

```python
E_s_d = (80*70) / 200
E_s_nd = (80*130) / 200
E_ns_d = (120*70) / 200
E_ns_nd = (120*130) / 200

print("Expected Frequencies of (Smoker, Disease) = ", E_s_d)
print("Expected Frequencies of (Smoker, No Disease) = ", E_s_nd)
print("Expected Frequencies of (No Smoker, Disease) = ", E_ns_d)
print("Expected Frequencies of (No Smoker, No Disease) = ", E_ns_nd)
```

#### step2 : Compute Chi-Square Statistic

```python
chi_sq = (50-E_s_d)**2/E_s_d + (30-E_s_nd)**2/E_s_nd + (20-E_ns_d)**2/E_ns_d + (100-E_ns_nd)**2/E_ns_nd

print("Chi-Square = ", round(chi_sq,2))
```

#### step3 : Determine Degrees of Freedom

```python
df = (2-1) * (2-1)
print("Degrees of Freedom:", df)
```

#### step4 : Interpret the Result

```python
'''
Using a chi-square distribution table, we compare the calculated chi-square value (44.33) 
with the critical value at one degree of freedom and a significance level (e.g., 0.05), 
approximately 3.841. Since 44.33 > 3.841, we reject the null hypothesis. 
This indicates a significant association between smoking status and the incidence 
of lung disease in this sample.
'''
```
